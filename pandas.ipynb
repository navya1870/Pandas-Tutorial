# Pivot Table

Pivot Table = isme jo khud ko dataframe bnana chahte ho from a given dataframe by providing valus columns and index uske basis pr create hoti h

```python
data = {
    'Product' : ['A','B','C','D'] * 5 ,
    'Region' : ['East','West','North','South'] * 5 ,
    'Sales' : np.random.randint(100,1000,20),
    'Units' : np.random.randint(10,100,20)
}
df = pd.DataFrame(data)
df
```

```python
pivot = pd.pivot_table(df,index = 'Product',columns = 'Region',values = 'Sales',aggfunc='median')
pivot
#aff cunction me default mean rehta h but mne isme median lga diya
```

```python
pivot1 = pd.pivot_table(df,index = 'Region', values = ['Sales','Units'],columns = 'Product')
pivot1
#values me hmesha quantitative data hi aaega cuz vrna agg function hi failed ho jaega categorical data me to aggregation function nikl hi ni skta
```

Cross tab

difference in cross tab and pivot table is that in pivot table we use aggregation functions while in cross tab we use counting functions

```python
pd.crosstab(df['Product'],df['Region'])
```

```python
import numpy as np
import pandas as pd
df = pd.read_csv('anime.csv')
df.head()
```

```python
# 1. make a new column for episode count
def extract_eps(txt):
  check = False
  data = ""
  for i in txt:
    if i == ')':
      break
    if check == True:
      data += i
    if i == '(':
      check = True
  return data
```

```python
df['Episode'] =  df['Title'].apply(extract_eps)
```

```python
df.head()
```

```python
# we need to remove that eps from all values so that episodes become a quantitative data
df['Episode'] = df['Episode'].str.replace(' eps','')
```

```python
df.head()
```

```python
#also need to convert these episodes into integer data type cuz it looks like interger but still it is string
type(df['Episode'][0])
```

```python
df['Episode'] = df['Episode'].astype(int)
```

now make a new column for time stamp

```python
df.iloc[0]['Title']
```

```python
#there is a time stamp that is months that are given now extract that
#hrr ek me same no. of space occupy hua h mtlb apr = 3 ,2009 = 4 ,jul = 3 and then 2010 = 4 and there is also space so total = 19
def extract_time(txt):
  check = False
  data = ""
  for i in range(len(txt)):
    if txt[i] == ")":
      for j in range(i+1,i+20):
        data += txt[j]
      break
  return data
```

```python
df['Time_stamp'] = df["Title"].apply(extract_time)
```

```python
df.head()
```

```python
df['Time_stamp'] = df['Time_stamp'].astype(str)

start = pd.to_datetime(df['Time_stamp'].str.split(' - ').str[0], format='%b %Y')
end = pd.to_datetime(df['Time_stamp'].str.split(' - ').str[1], format='%b %Y')

df['Total_Months'] = (end.dt.year - start.dt.year) * 12 + (end.dt.month - start.dt.month)
```
